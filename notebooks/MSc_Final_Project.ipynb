# Civil Infrastructure Damage Detection Using YOLOv4

## MSc Final Project

This project focuses on detecting potholes in road images using deep learning and computer vision. A YOLOv4 object detection model was configured and trained to identify potholes from road-surface imagery.

### Project Objectives

- Prepare and organise road-image data for object detection.
- Configure the YOLOv4 architecture for pothole detection.
- Train the model using annotated road images.
- Evaluate the trained model on unseen images.
- Detect and localise potholes using bounding boxes.
